# Key Insights & Business Recommendations — Olist E-Commerce Analysis

This notebook consolidates the most significant findings from the full analysis 
(`01_data_cleaning.ipynb` through `04_eda_delivery.ipynb`) into actionable business insights.

## 1. Business Context

Olist is a Brazilian e-commerce platform connecting small/medium sellers to major marketplaces. 
This analysis covers ~99,000 orders placed between September 2016 and September 2018, examining 
sales trends, product and customer behavior, and delivery performance to identify what drives 
revenue and customer satisfaction — and where the business has room to improve.

## 2. Key Insights

### Growth & Seasonality
- Revenue grew steadily through 2017, peaked sharply in **November 2017 (R 1.01M)** — almost 
  certainly Black Friday — then plateaued at **R 850K-1M/month** through mid-2018. The business 
  matured from a growth phase into a stable-volume phase within ~18 months.
- November is a clear seasonal peak (~40% above average month). Outside this, demand is fairly 
  stable year-round with a modest January-February post-holiday dip.

### Revenue Concentration
- **São Paulo dominates**: 2.8x the revenue of the next state (RJ), and the top 5 states generate 
  ~74% of total revenue. This concentration is driven by **order volume, not higher spend per 
  order** — SP actually has one of the lowest average order values among top states.
- The customer base mirrors this: São Paulo city alone accounts for **15.6%** of all customers 
  nationwide.

### Product Performance
- Revenue is moderately concentrated: the top 10 of 74 categories drive ~62% of revenue 
  (`health_beauty`, `watches_gifts`, `bed_bath_table` lead).
- `watches_gifts` stands out as a rare category combining **both** high volume and premium pricing 
  — a particularly high-value category worth prioritizing in marketing/inventory decisions.

### Customer Behavior
- Only **3.05%** of customers made repeat purchases, but they spend **~2x more on average** 
  (R 262 vs R 139). Repeat customers still contribute just ~5.6% of revenue due to their small 
  numbers — meaning the business currently relies overwhelmingly on new customer acquisition, 
  not retention.

### Delivery & Satisfaction — the strongest link found
- **Late delivery is the single strongest driver of dissatisfaction** found in this analysis: 
  review scores collapse from **4.29 (on-time) to 2.27 (late)** — far outweighing any other 
  factor examined, including category or payment method.
- Delivery reliability varies sharply by region: Brazil's North region (RR, AP, AM) sees 
  26-29 day average deliveries, and specifically **Alagoas (AL)** stands out as both slow 
  *and* unreliable (78.6% on-time) — a genuine logistics weak point distinct from other slow 
  regions that remain reliable despite distance.
- Furniture categories (e.g. `office_furniture`, 3.49 avg review) rate poorly, but this was 
  **ruled out** as a delivery-speed issue (91.9% on-time, near-average) — pointing instead to 
  product-related causes like assembly difficulty or in-transit damage.

## 3. Business Recommendations

1. **Prioritize on-time delivery as a satisfaction lever.** The 4.29 vs 2.27 review score gap 
   between on-time and late orders is the largest effect found in this entire analysis — larger 
   than product category, payment method, or price. Investing in delivery reliability likely has 
   the highest ROI for improving customer satisfaction.

2. **Investigate logistics specifically in Alagoas (AL).** Unlike other slow-delivery regions 
   (e.g. Amazonas) which remain reliable despite long transit times, AL is both slow *and* 
   unreliable — suggesting a fixable local logistics issue rather than an unavoidable geographic 
   constraint.

3. **Investigate `office_furniture` quality/fulfillment, not shipping.** Since delivery speed is 
   not the cause of its low ratings, the issue likely lies in product packaging, assembly 
   instructions, or damage in transit — worth a focused seller/quality review rather than a 
   logistics fix.

4. **Build a retention strategy.** Repeat customers spend ~2x more, yet only 3% of customers 
   return. Even a modest improvement in repeat rate (e.g. through post-purchase engagement or 
   loyalty incentives) could meaningfully shift revenue mix toward higher-value customers.

5. **Prepare inventory and marketing for November demand** given the consistent, sharp seasonal 
   spike — likely Black Friday — to avoid stockouts or delivery slippage during the highest-demand 
   period (which would be especially costly given finding #1).

## 4. Limitations

- Dataset spans ~2 years (Sep 2016 - Sep 2018), a relatively short window to fully assess 
  repeat-purchase behavior, especially for infrequently-bought categories.
- Review comments (text) were not analyzed — only numeric scores — so the specific reasons behind 
  furniture's low ratings remain a hypothesis, not confirmed.
- A small number of source data inconsistencies were identified and documented (e.g. 8 "delivered" 
  orders missing a delivery date, 1 order missing payment data) — negligible in scale but noted 
  for transparency.
